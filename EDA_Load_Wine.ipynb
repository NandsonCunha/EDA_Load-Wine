```python
import matplotlib.pyplot as plt

f1 = 'malic_acid'
f2 = 'total_phenols'

fig = plt.figure(figsize = (6, 3))

axes =fig.add_axes([0, 0, 1, 1])

c = {0: 'r',
     1: 'b',
     2: 'g'}

for specie in y.unique():
  axes.scatter(X[y==specie][f1], X[y==specie][f2], c = c[specie])


axes.set_xlabel(f1)
axes.set_ylabel(f2);
```
